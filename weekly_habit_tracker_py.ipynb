```python
DAYS = ("Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun")
habit = input("Habit to track: ")
checklist = tuple(input(f"Done on {day}? (y/n): ").lower() == 'y' for day in DAYS)
print(f"\n--- {habit} Report ---")
print(f"Habit: {habit}")
[print(f"{day}: {'y' if done else 'n'}") for day, done in zip(DAYS, checklist)]
print(f"Score: {checklist.count(True)}/7 ({checklist.count(True)/7:.1%})")
```
