```python
 num = int(input("Enter the number of subjects: "))

# Create an empty list for marks
marks = []

# Collect marks using a loop
for i in range(n):
    score = float(input(f"Enter mark for subject {i + 1}: "))
    marks.append(score)

# Analyze the marks
total = sum(marks)
average = total / n
highest = max(marks)
lowest = min(marks)

# Determine the grade based on average
if average >= 90:
    grade = "A"
elif average >= 80:
    grade = "B"
elif average >= 70:
    grade = "C"
elif average >= 60:
    grade = "D"
else:
    grade = "F"

# Print the analysis report
print("\n--- Mark List Analysis Report ---")
print(f"Marks entered: {marks}")
print(f"Total Marks: {total}")
print(f"Average Mark: {average:.2f}")
print(f"Highest Mark: {highest}")
print(f"Lowest Mark: {lowest}")
print(f"Overall Grade: {grade}")
```
